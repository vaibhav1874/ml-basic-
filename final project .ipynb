```python
from scipy.stats import pearsonr


selected_features = [
    'Age', 'is_female', 'RestingBP', 'Cholesterol', 'FastingBS', 'MaxHR',
       'is_ExerciseAngina', 'Oldpeak', 'is_up', 'HeartDisease',
       'ChestPainType_ASY', 'ChestPainType_ATA', 'ChestPainType_NAP',
       'ChestPainType_TA', 'RestingECG_LVH', 'RestingECG_Normal',
       'RestingECG_ST', 'RestingBP_cat_Normal', 'RestingBP_cat_Elevated',
       'RestingBP_cat_High', 'RestingBP_cat_Very High'
]

correlation={

   feature:pearsonr(df1[feature],df1['HeartDisease'])[0]
   for feature  in selected_features
    
}

corr_df=pd.DataFrame(list(correlation.items()), columns=['feature','personal correlation'])
corr_df.sort_values(by='personal correlation',ascending=False)
```

```python
cat_feature=[ 'Age', 'is_female', 'RestingBP', 'Cholesterol', 'FastingBS', 'MaxHR',
       'is_ExerciseAngina', 'Oldpeak', 'is_up', 'HeartDisease',
       'ChestPainType_ASY', 'ChestPainType_ATA', 'ChestPainType_NAP',
       'ChestPainType_TA', 'RestingECG_LVH', 'RestingECG_Normal',
       'RestingECG_ST', 'RestingBP_cat_Normal', 'RestingBP_cat_Elevated',
       'RestingBP_cat_High', 'RestingBP_cat_Very High']
```

```python
from scipy.stats import chi2_contingency
import pandas as pd

alpha = 0.05

# Divide charges into 4 equal-frequency bins
df1['HeartDisease_bin'] = pd.qcut(
    df1['HeartDisease'],
    q=4,
    labels=False,
    duplicates='drop'
)

chi2_results = {}

for col in cat_feature:

    # Create contingency table
    contingency = pd.crosstab(
        df1[col],
        df1['HeartDisease_bin']
    )

    # Chi-Square test
    chi2_stat, p_val, _, expected = chi2_contingency(contingency)

    # Decision
    decision = (
        'Reject Null (Keep Feature)'
        if p_val < alpha
        else 'Fail to Reject Null (Drop Feature)'
    )

    chi2_results[col] = {
        'chi2_statistic': chi2_stat,
        'p_value': p_val,
        'Decision': decision
    }

# Display results
results_df = pd.DataFrame(chi2_results).T

print(results_df)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score,f1_score,classification_report
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
```

```python
X = df1.drop(columns=['HeartDisease', 'HeartDisease_bin'], errors='ignore')
y=df1['HeartDisease']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)  
```

```python
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train) 
X_test_scaled = scaler.transform(X_test)        
```

```python
models = {
    "logistic regression": LogisticRegression(),
    "decision tree": DecisionTreeClassifier(),
    "gaussian nb": GaussianNB(),
    "svm": SVC(),
    "k neighbors": KNeighborsClassifier()
}
result=[]
```

```python
for name,model in models.items():
    model.fit(X_train_scaled,y_train)
    y_pred=model.predict(X_test_scaled)
    acc=accuracy_score(y_test,y_pred)
    f1=f1_score(y_test,y_pred)

    result.append({

        'model':name,
        'accuracy':round(acc,4),
        'f1_score':round(f1,4)
    })
```
