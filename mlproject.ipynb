```python

x=df.drop(columns=['price'],axis=1)
y=df['price']
```

```python
x
```

```python
x_ne_encode=pd.get_dummies(x,columns=['model','transmission','fuelType']).astype(int)
```

```python
x_ne_encode
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
encode=LabelEncoder()
```

```python
columns=['model','transmission','fuelType']
xlable=x
```

```python
for i in columns:
   xlable[i]= encode.fit_transform(xlable[i])
```

```python
xlable
```

```python
from sklearn.preprocessing import StandardScaler
numeric_col=['year','mileage','tax','mpg']
scaler=StandardScaler()
x_ne_encode[numeric_col]= scaler.fit_transform(x_ne_encode[numeric_col])
```

```python
x_ne_encode
```

```python
xlable.columns
```

```python
xlable[['model', 'year', 'transmission', 'mileage', 'fuelType', 'tax', 'mpg','engineSize']]=scaler.fit_transform(xlable[['model', 'year', 'transmission', 'mileage', 'fuelType', 'tax', 'mpg','engineSize']])
```

```python
xlable
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
```

```python
X_train, X_test, y_train, y_test = train_test_split(  
   x_ne_encode, y, test_size=0.33, random_state=42)  
```

```python
model=LinearRegression()
model.fit(X_train,y_train)
```

```python
y_predict= model.predict(X_test)
y_predict
```

```python
y_test
```

```python
r2=r2_score(y_test,y_predict)
```

```python
r2
```

```python
n=X_test.shape[0]
p=X_test.shape[1]
adjuste_r2=1 - ((1 - r2) * (n - 1) / (n - p - 1))

```

```python
adjuste_r2
```
